```python
import re

positive_words = {"great", "good", "love", "happy", "recommended"}
negative_words = {"terrible", "bad", "hate", "sad", "poor"}

def analyze_sentiment(text):
    words = re.findall(r"[a-zA-Z']+", text.lower())

    positive_count = sum(1 for word in words if word in positive_words)
    negative_count = sum(1 for word in words if word in negative_words)

    score = positive_count - negative_count

    if score > 0:
        label = "Positive"
    elif score < 0:
        label = "Negative"
    else:
        label = "Neutral"

    return score, label

examples = [
    "This is great",
    "This is terrible",
    "This is not great",
    "The food is great but the service is terrible"
]

for text in examples:
    score, label = analyze_sentiment(text)
    print(f"{text} → Score: {score}, Label: {label}")
```

```python
def analyze_with_negation(text):
    words = re.findall(r"[a-zA-Z']+", text.lower())
    score = 0

    for i, word in enumerate(words):
        word_score = 0

        if word in positive_words:
            word_score = 1
        elif word in negative_words:
            word_score = -1

        if i > 0 and words[i - 1] == "not":
            word_score = -word_score

        score += word_score

    if score > 0:
        label = "Positive"
    elif score < 0:
        label = "Negative"
    else:
        label = "Neutral"

    return score, label


for text in ["This is great", "This is not great", "This is not bad"]:
    score, label = analyze_with_negation(text)
    print(f"{text} → Score: {score}, Label: {label}")
```

```python
test_sentences = [
    "I love this",
    "This is not good",
    "The service is not terrible",
    "This is not very great",
    "The food is great but the service is terrible",
    "This is not recommended",
"I am not happy"
]

for text in test_sentences:
    old_score, old_label = analyze_sentiment(text)
    new_score, new_label = analyze_with_negation(text)

    print(text)
    print(f"  Original: {old_label} ({old_score})")
    print(f"  Improved: {new_label} ({new_score})")
    print()
```

```python
import pandas as pd

results = []

for text in test_sentences:
    old_score, old_label = analyze_sentiment(text)
    new_score, new_label = analyze_with_negation(text)

    results.append({
        "Text": text,
        "Original": old_label,
        "Improved": new_label,
        "Original score": old_score,
        "Improved score": new_score
    })

comparison = pd.DataFrame(results)
display(comparison)

comparison.to_csv("sentiment_comparison.csv", index=False)
```
